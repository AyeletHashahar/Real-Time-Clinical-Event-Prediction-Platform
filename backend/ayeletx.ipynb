```python
import csv

def txt_to_csv(txt_path, csv_path):
    with open(txt_path, 'r') as f:
        lines = f.readlines()

    # Ignore the first two lines
    lines = lines[2:]

    data = []
    current_entity = None

    for line in lines:
        line = line.strip()
        if not line:
            continue
        if line.endswith(';') and ',' not in line:
            # This is an entity ID line
            current_entity = int(line[:-1])
        else:
            # This is a data line
            records = line.split(';')
            for record in records:
                record = record.strip()
                if not record:
                    continue
                fields = record.split(',')
                if len(fields) == 4:
                    start, end, symbolid, temporalpropertyid = fields
                    data.append([current_entity, int(start), int(end), int(symbolid), int(temporalpropertyid)])

    # Write to CSV
    with open(csv_path, 'w', newline='') as f:
        writer = csv.writer(f)
        writer.writerow(['entity_id', 'start', 'end', 'symbolid', 'temporalpropertyid'])
        writer.writerows(data)
```
